# Task 1 – Data Cleaning and Preprocessing
## Customer Personality Analysis

**Objective:** Clean the raw `marketing_campaign.csv` dataset by checking missing values, duplicates, inconsistent text, date formats, column names, and data types.

**Dataset:** Customer Personality Analysis (Kaggle). The source file is tab-separated (`.csv` extension but tab delimiter), with 2,240 records and 29 columns in the commonly distributed version.

**Important:** Upload the original `marketing_campaign.csv` file when prompted. This notebook does not fabricate a cleaned output; it creates the cleaned CSV from the file you supply.

## 1. Import libraries

```python
import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

## 2. Load the raw dataset
The original dataset is commonly tab-delimited. The code below detects the separator automatically. In Google Colab, it opens an upload dialog; in local Jupyter, enter the file path.

```python
try:
    from google.colab import files
    uploaded = files.upload()
    input_file = next(iter(uploaded))
except ImportError:
    input_file = input("Enter path to marketing_campaign.csv: ").strip()

# sep=None with the Python engine detects tabs/commas in most cases.
df_raw = pd.read_csv(input_file, sep=None, engine="python", encoding="utf-8-sig")

print("Raw dataset shape:", df_raw.shape)
display(df_raw.head())
```

## 3. Understand the dataset
Inspect dimensions, data types, sample values, and summary statistics before changing anything.

```python
print("Rows and columns:", df_raw.shape)
print("\nData types and non-null counts:")
df_raw.info()

print("\nDescriptive statistics:")
display(df_raw.describe(include="all").T)

print("\nOriginal column names:")
print(df_raw.columns.tolist())
```

## 4. Standardize column names
Convert headers to lowercase `snake_case`, remove leading/trailing spaces, and replace spaces or punctuation with underscores.

```python
import re

df = df_raw.copy()

def clean_column_name(name):
    name = str(name).strip().lower()
    name = re.sub(r"[^a-z0-9]+", "_", name)
    return name.strip("_")

df.columns = [clean_column_name(c) for c in df.columns]
print(df.columns.tolist())
```

## 5. Check and handle missing values
First measure missing values. Blank strings are treated as missing. For this dataset, `income` is numeric and is imputed with its median because the median is less sensitive to extreme income values than the mean. Other columns are not filled blindly.

```python
# Treat empty/whitespace-only text cells as missing.
df = df.replace(r"^\s*$", np.nan, regex=True)

missing_before = df.isna().sum().sort_values(ascending=False)
missing_report = pd.DataFrame({
    "missing_count": missing_before,
    "missing_percent": (missing_before / len(df) * 100).round(2)
})
display(missing_report[missing_report["missing_count"] > 0])

if "income" in df.columns:
    df["income"] = pd.to_numeric(df["income"], errors="coerce")
    income_median = df["income"].median()
    df["income"] = df["income"].fillna(income_median)
    print(f"Filled missing income values with median: {income_median:,.2f}")
else:
    print("Column 'income' not found; no income imputation performed.")
```

## 6. Remove duplicate records
Remove exact duplicate rows. We do not automatically remove repeated IDs unless they are confirmed to be duplicate customer records; inspect ID duplicates separately.

```python
exact_duplicates = int(df.duplicated().sum())
print("Exact duplicate rows found:", exact_duplicates)

df = df.drop_duplicates().copy()

if "id" in df.columns:
    duplicate_ids = int(df["id"].duplicated().sum())
    print("Repeated ID values after exact-row removal:", duplicate_ids)
    if duplicate_ids:
        print("Review repeated IDs before removing: they may represent data issues or multiple records.")
```

## 7. Standardize text columns
Trim whitespace and normalize capitalization. We preserve distinct source categories (for example, `Together` is not changed to `Married`) because they may represent different categories in the source data.

```python
text_columns = df.select_dtypes(include=["object", "string"]).columns.tolist()

for col in text_columns:
    df[col] = df[col].astype("string").str.strip()

# Normalize case for known categorical fields without merging categories.
for col in ["education", "marital_status"]:
    if col in df.columns:
        df[col] = df[col].str.replace(r"\s+", " ", regex=True).str.title()

for col in text_columns:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False).head(20))
```

## 8. Convert data types and standardize dates
`dt_customer` is parsed as a date using day-first interpretation (the source commonly uses `dd-mm-yyyy`). Invalid dates become `NaT` and are counted for review. Numeric columns are converted safely.

```python
if "dt_customer" in df.columns:
    df["dt_customer"] = pd.to_datetime(
        df["dt_customer"], dayfirst=True, errors="coerce"
    )

# Convert known numeric fields where present.
numeric_candidates = [
    "id", "year_birth", "income", "kidhome", "teenhome", "recency",
    "mntwines", "mntfruits", "mntmeatproducts", "mntfishproducts",
    "mntsweetproducts", "mntgoldprods", "numdealspurchases",
    "numwebpurchases", "numcatalogpurchases", "numstorepurchases",
    "numwebvisitsmonth", "acceptedcmp1", "acceptedcmp2", "acceptedcmp3",
    "acceptedcmp4", "acceptedcmp5", "complain", "z_costcontact",
    "z_revenue", "response"
]

for col in numeric_candidates:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

print(df.dtypes)
if "dt_customer" in df.columns:
    print("Unparseable/missing customer dates:", int(df["dt_customer"].isna().sum()))
```

## 9. Validate values and review potential outliers
Outliers are flagged for investigation, not automatically deleted. Extreme values can be genuine customers and removing them without business context can distort the data.

```python
# Basic domain checks (flag only; do not silently delete records).
checks = {}

if "year_birth" in df.columns:
    checks["year_birth_outside_1900_2014"] = int(
        (~df["year_birth"].between(1900, 2014) & df["year_birth"].notna()).sum()
    )

if "income" in df.columns:
    checks["negative_income"] = int((df["income"] < 0).sum())

if "response" in df.columns:
    checks["response_not_0_or_1"] = int(
        (~df["response"].isin([0, 1]) & df["response"].notna()).sum()
    )

print("Validation flags:", checks)

# IQR outlier count for income (if available).
if "income" in df.columns:
    q1, q3 = df["income"].quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    outlier_count = int(((df["income"] < lower) | (df["income"] > upper)).sum())
    print(f"Income IQR outlier candidates: {outlier_count} (retained for review)")
```

## 10. Final quality checks

```python
print("Cleaned dataset shape:", df.shape)
print("\nRemaining missing values:")
display(df.isna().sum().sort_values(ascending=False).to_frame("missing_count"))

print("\nRemaining exact duplicates:", int(df.duplicated().sum()))
print("\nCleaned sample:")
display(df.head())
```

## 11. Export cleaned CSV and cleaning summary
Dates are saved in ISO format (`YYYY-MM-DD`) for consistency. The CSV is exported to the notebook's current working directory and downloaded automatically in Colab.

```python
output_file = "cleaned_customer_personality_analysis.csv"
df.to_csv(output_file, index=False, date_format="%Y-%m-%d")

summary = {
    "raw_rows": len(df_raw),
    "raw_columns": len(df_raw.columns),
    "exact_duplicates_removed": exact_duplicates,
    "cleaned_rows": len(df),
    "cleaned_columns": len(df.columns),
    "missing_values_remaining": int(df.isna().sum().sum()),
    "output_file": output_file
}
print("Cleaning summary:")
for key, value in summary.items():
    print(f"{key}: {value}")

try:
    from google.colab import files
    files.download(output_file)
except ImportError:
    print(f"Saved locally as: {Path(output_file).resolve()}")
```
